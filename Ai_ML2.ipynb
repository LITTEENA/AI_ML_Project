```python
pd.crosstab(data.target,data.sex)
```

```python
pd.crosstab(data.target,data.sex).plot(kind = "bar" , figsize = (10,6),color=['indigo','orange'])
plt.title("Heart Disease summary sex wise")
plt.xlabel("0 = no heart disese , 1 = Heart disease")
plt.ylabel("No. of people")
plt.legend(['Female','Male'])
```

```python
plt.figure(figsize = (10,6))
plt.scatter(data.age[data.target ==1],data.thalach[data.target ==1],c='red')
plt.scatter(data.age[data.target ==0],data.thalach[data.target ==0],c='green')
plt.title("Age vs Heart Rate")
plt.xlabel("Age")
plt.ylabel("thalach")
plt.legend(['Heart Problem','No Heart Problem'])
```
